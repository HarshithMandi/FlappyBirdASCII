```python
import itertools as itt

for i in itt.count(10,5):
    if i>30:
        break
    print(i)
```

```python
count=0
for i in itt.cycle("ABCD"):
    if count>10:
        break
    count+=1
    print(i)
```

```python
for i in itt.repeat("Hello",3):
    print(i)
    
```

```python
for i in itt.permutations("ABC",2):
    print(i)
```

```python
for i in itt.combinations("ABC",2):
    print(i)
```
