```python
add=lambda a,v:a+v
print(add(10,20))
```

```python
l1=[lambda a:a+10,lambda a:a+20,lambda a:a+30]
for i in l1:
    print(i(10))
```

```python
logs=[
    "INFO: Server started",
    "ERROR: Database connection failed",
    "WARNING: Low disk space",
    "INFO: User logged in",
    "ERROR: File not found",
]
errors=list(filter(lambda x: "ERROR" in x,logs))
errors
```

```python
#sorts tasks based on priority level

tasks=[
    ("Backup",3),
    ("Security scan",1),
    ("Software update",2),
]
tasks.sort(key=lambda x: x[1])
tasks
```

```python
from functools import reduce

data=[10,20,30,40,50,60]

data2=list(map(lambda x:x*2,data))
print(data2)
data3=list(filter(lambda x:x>40,data2))
print(data3)
total=reduce(lambda a,b:a+b,data3)
print(total)
```
