```python
# order processing

# user_input = input("Enter a list of tuples (CAUTION: use trusted input): ")
# try:
#     list_of_tuples = [eval(user_input)]
#     print(f"Successfully parsed list of tuples: {list_of_tuples}")
# except (NameError, SyntaxError):
#     print("Invalid input.")

list_of_tuples=[("Alice","Laptop"),("Bob","Mobile"),("Alice","Mouse"),("David","Mobile"),("Bob","Laptop"),("Alice","Mobile")]
electronics={"Laptop":50000,"Mobile":20000}
accessories={"Mouse":500,"Keyboard":1000}

distinct_orders={}

for (u,v) in list_of_tuples:
    distinct_orders[u]=[]

distinct_orders

for (u,v) in list_of_tuples:
    distinct_orders[u].append(v)

distinct_orders

print("product sales")
for (u,v) in list_of_tuples:
    if v in electronics:
        print(f"{v} sold for {electronics[v]}")
    elif v in accessories:
        print(f"{v} sold for {accessories[v]}")

print("Orders grouped by customer:")
for customer, items in distinct_orders.items():
    itemstring=", ".join(items)
    print(f"  {customer}-> {itemstring}")
```

```python
#iterators, generators and decorators

numbers=[10,20,30,40,50]

curr=iter(numbers)
print(curr)
print(next(curr))
print(next(curr))
print(next(curr))
print(next(curr))
```

```python


# Create an iterator that cycles through a list indefinitely
cycler = itertools.cycle(['a', 'b', 'c'])

# Get the first 10 elements from the cycler
result = [next(cycler) for _ in range(10)]
print(*result, sep=', ')
```

```python
def numbers():
    for i in range(1,6):
        yield i
x=numbers()
print(x)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```
